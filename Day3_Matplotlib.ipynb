Matplotlib is a Python library used to create:
Line Charts
Bar Charts
Pie Charts
Scatter Plots
Histograms
Box Plots
Area Charts
Heatmaps (with other libraries)
It helps you visualize data, making patterns and trends easier to understand.

# ***Your First Graph***

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,15,30,25]

plt.plot(x, y)
plt.show()
```

# **Line Chart**

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [100,150,120,180,200]

plt.plot(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Change Line Style
plt.plot(
    months,
    sales,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o"
)
```

# **Bar Chart**

```python
import matplotlib.pyplot as plt

students = ["Vikky","Rahul","Aman","Neha"]
marks = [85,90,78,95]

plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#Horizontal Bar Chart
plt.barh(students, marks)
plt.show()
```

# **Pie Chart**

```python
import matplotlib.pyplot as plt

subjects = ["AI","Python","Java","SQL"]
hours = [35,25,20,20]

plt.pie(hours, labels=subjects, autopct="%1.1f%%")
plt.title("Study Time")
plt.show()
```

# **Scatter Plot**

```python
import matplotlib.pyplot as plt

height = [150,155,160,165,170]
weight = [50,55,60,65,70]

plt.scatter(height, weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

# **Histogram**

```python
#Shows how data is distributed.
import matplotlib.pyplot as plt

marks = [55,60,65,70,75,80,85,90,95]

plt.hist(marks, bins=5)
plt.title("Marks Distribution")
plt.show()
```

# **Box Plot**

```python
"""
Useful for finding:
Median
Quartiles
Outliers
"""

import matplotlib.pyplot as plt

marks = [50,55,60,70,75,80,85,90,100]

plt.boxplot(marks)
plt.title("Marks")
plt.show()
```

# **Grid**

```python
#Adds grid lines to improve readability.
plt.plot(x, y)
plt.grid(True)
plt.show()
```
