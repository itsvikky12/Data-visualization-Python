```python
#Remove Columns
data.drop("Cabin", axis=1)
```

```python
#Permanently remove:
data.drop("Cabin", axis=1, inplace=True)
```

```python
#Multiple columns:
data.drop(["Cabin", "Ticket"], axis=1, inplace=True)
```

```python
#Rename one column:
data.rename(columns={"Fare": "Ticket_Fare"}, inplace=True)
```

```python
#Rename Columns
data.rename(columns={
    "PassengerId": "Passenger_ID",
    "Survived": "Survival"
}, inplace=True)
```

```python
#Add New Column
data["FamilySize"] = data["SibSp"] + data["Parch"] + 1
```

```python
data["Survived"] = data["Survived"].map({
    0: "No",
    1: "Yes"
})
```

```python
data["Sex"] = data["Sex"].map({
    "male": 0,
    "female": 1
})
```

```python
def age_group(age):
    if age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

data["AgeGroup"] = data["Age"].apply(age_group)
```

```python
data["AgeGroup"] = data["Age"].apply(
    lambda x: "Child" if x < 18 else "Adult"
)
```
