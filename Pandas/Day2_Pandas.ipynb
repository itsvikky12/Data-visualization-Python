# **Pandas**

# What is Pandas?
Pandas is a Python library used for data manipulation and analysis. It makes it easy to work with structured data such as CSV files, Excel sheets, SQL tables, and JSON data.

With Pandas, you can:
   1. Read CSV or Excel files
   2. Filter rows
   3. Select columns
   4. Calculate averages
   5. Handle missing values
   6. Sort data
   7. Group data
   8. Merge multiple datasets
   9. Prepare data for Machine Learning
